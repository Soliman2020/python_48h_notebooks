# Part 1 — Git: a practical introduction

## What is Git?

Git is a version-control system. It records changes to files so that you can see what changed, return to an earlier commit, and describe the history of a project.

- **Repository:** the project folder and its Git history.
- **Working tree:** the files you are currently editing.
- **Staging area:** the changes selected for the next commit.
- **Commit:** a named snapshot of staged changes.
- **GitHub:** an optional online service that can store a remote copy of a Git repository. Git and GitHub are not the same thing.

## The basic local workflow

Run these commands in a terminal opened in your project folder:

```text
git status                 # What has changed?
git diff                   # What are the line-by-line changes?
git add filename.py       # Select a file for the next commit
git commit -m "Describe the change"
git log --oneline          # Show short commit history
```

The normal cycle is: **edit → status/diff → add → commit**. Commit small, meaningful steps.

You will analyse a supplied sales CSV. The goal is simple exploratory data analysis, not advanced statistics. You will inspect the data, calculate revenue, summarize by product and region, make charts, and explain what you found.

The file is expected at `datasets/student_sales.csv`. If you are working in a copy of the notebook without the file, run the setup cell below once to create the supplied practice file.

````python
from pathlib import Path

data_folder = Path("datasets")
data_folder.mkdir(exist_ok=True)
csv_path = data_folder / "student_sales.csv"

if not csv_path.exists():
    csv_path.write_text("""date,region,product,units,unit_price
2026-01-05,North,Notebook,12,4.50
2026-01-06,South,Pen,30,1.25
2026-01-07,East,Folder,18,2.75
2026-01-08,West,Notebook,9,4.50
2026-01-09,North,Pen,25,1.25
2026-01-10,South,Folder,14,2.75
2026-01-11,East,Notebook,16,4.50
2026-01-12,West,Pen,22,1.25
2026-01-13,North,Folder,11,2.75
2026-01-14,South,Notebook,20,4.50
2026-01-15,East,Pen,28,1.25
2026-01-16,West,Folder,13,2.75
2026-01-17,North,Notebook,15,4.50
2026-01-18,South,Pen,27,1.25
2026-01-19,East,Folder,10,2.75
2026-01-20,West,Notebook,8,4.50
""", encoding="utf-8")

print(csv_path, csv_path.exists())
````

## Step 1 — load and inspect the CSV

Use `read_csv()`, then inspect the first rows, dimensions, column names, data types, and summary information.

````python
import pandas as pd

sales = pd.read_csv(csv_path)
display(sales.head())
print("Shape:", sales.shape)
print("Columns:", list(sales.columns))
sales.info()
````

## Step 2 — check data quality

Check missing values, duplicated rows, and numeric summary statistics. Do not assume a dataset is clean just because it loaded successfully.

````python
print("Missing values:")
display(sales.isna().sum())
print("Duplicate rows:", sales.duplicated().sum())
display(sales.describe())
````

## Step 3 — create a calculated column

Revenue is `units * unit_price`. Make a copy before adding analysis columns so that the original loaded data remains clear.

````python
analysis = sales.copy()
analysis["date"] = pd.to_datetime(analysis["date"])
analysis["revenue"] = analysis["units"] * analysis["unit_price"]
display(analysis.head())
print("Total revenue:", analysis["revenue"].sum())
````

## Step 4 — filter and group

Answer these questions with Pandas:

1. Which rows have at least 15 units?
2. What is the total revenue for each product?
3. What is the total revenue for each region?
4. Which product generated the most revenue?

````python
high_volume = analysis[analysis["units"] >= 15]
revenue_by_product = analysis.groupby("product")["revenue"].sum().sort_values(ascending=False)
revenue_by_region = analysis.groupby("region")["revenue"].sum().sort_values(ascending=False)

display(high_volume)
display(revenue_by_product)
display(revenue_by_region)
````

## Step 5 — make simple charts

Charts should answer a question. Use one chart for product revenue and one for regional revenue. Add a title and axis labels.

````python
import matplotlib.pyplot as plt

revenue_by_product.plot(kind="bar", title="Revenue by Product", ylabel="Revenue", xlabel="Product")
plt.tight_layout()
plt.show()

revenue_by_region.plot(kind="bar", title="Revenue by Region", ylabel="Revenue", xlabel="Region")
plt.tight_layout()
plt.show()
````
