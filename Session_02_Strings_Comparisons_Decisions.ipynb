```python
# String
name = "Soliman"
multi_line_string_as_comment = """This is a
multi-line string."""

print(name)
print(multi_line_string_as_comment)
```

```python
# Basic String Operations

# Concatenation
str1 = "Hello"
str2 = "World"

concatenated = str1 + " " + str2

print("Concatenation: ", concatenated)
```

```python
# Repetition
repeated = str1 * 3
print("Repetition: ", repeated)
```

```python
# Slicing
sliced = concatenated[0:5]
print("Slicing: ", sliced)
```

```python
# Length
length = len(concatenated)
print("Length: ", length)
```

```python
# Membership
is_in = "World" in concatenated
print("Membership (in): ", is_in)
```

```python
# String Methods

# Upper case
upper_str = concatenated.upper()
print("Upper case: ", upper_str)
```

```python
# Lower case
lower_str = concatenated.lower()
print("Lower case: ", lower_str)
```

```python
# Capitalize
capitalized_str = concatenated.capitalize()
print("Capitalize: ", capitalized_str)
```

```python
# Title case
title_str = concatenated.title()
print("Title case: ", title_str)
```

```python
# Strip
strip_str = "   Hello World   ".strip()
print("Strip: ", strip_str)
```

```python
# Replace
replaced_str = concatenated.replace("World", "Python")
print("Replace: ", replaced_str)
```

```python
# Split
split_str = concatenated.split(" ")
print("Split: ", split_str)
```

```python
# Join
joined_str = "-".join(split_str)
print("Join: ", joined_str)
```

```python
# Find
found_index = concatenated.find("World")
print("Find: ", found_index)
```

```python
# Format
formatted_str = "Hello, {}!".format("Python")
print("Format: ", formatted_str)
```

```python
# Format _ 2nd method
var= 'partner'
formatted_str2 = f"Hello, {var}!"
print("Format: ", formatted_str2)
```

## Concept 2 — Comparisons and Boolean expressions

`==`, `!=`, `<`, `<=`, `>` and `>=` compare values. `and`, `or`, `not` combine conditions. A Boolean expression evaluates to `True` or `False`; it does not make a decision by itself.

**Predict → run → change:** Before running the next code cell, explain what it should print. Next, change **one** value and observe the result.

```python
priority='high'
hours_open=30
is_urgent=(priority=='high' and hours_open>24)
print('Escalate?',is_urgent)
print('Normal?',not is_urgent)
```

**Stop and think:** Explain why `priority = "high"` is not a comparison.

**Industry transfer:** The logic above can be applied to telecom device alerts; laboratory quality ranges; audit and hr routing. What is the equivalent of our variables or records in your work?

## Concept 3 — If / elif / else

Python uses indentation to group blocks. Always explain the decision in plain English first, then translate it into conditions. Branches run from top to bottom; only the first matching `if`/`elif` branch executes.

**Predict → run → change:** Before running the next code cell, explain what it should print. Next, change **one** value and observe the result.

```python
score=76
if score>=85:
    label='excellent'
elif score>=70:
    label='pass'
else:
    label='review'
print(label)
```

**Stop and think:** Try boundary values 69, 70, 84 and 85.
