## Concept 1 — For loops, totals and counters

A `for` loop repeats an indented block for each item. Start with a count or sum, then add a condition. `range(n)` generates 0 through n−1. `enumerate` pairs indices with values.

**Predict → run → change:** Before running the next code cell, explain what it should print. Next, change **one** value and observe the result.

### For Loop

```python
# For loops Examples
for i in range(5):
    print("Iteration:", i)

for item in [10, 20, 30]:
    print("ele:", item+2)
```

```python
durations=[2,5,9,3]
total=0
long_cases=0

for duration in durations:
    total+=duration
    if duration>4:
        long_cases+=1
print('Total hours:',total,'Long cases:',long_cases)

for position,value in enumerate(durations,start=1):
    print(position,value)
```

## Concept 3 — While loops and beginner comprehensions

A `while` loop keeps running while its condition is true. Always update the variable or define a safe exit. A list comprehension is an **optional shorthand** for a simple transformation; write a regular loop first.

**Predict → run → change:** Before running the next code cell, explain what it should print. Next, change **one** value and observe the result.

### While Loop

```python
# While loops Example
counter = 0

while counter <= 5:
    print("Count:", counter)
    counter += 1
```

```python
attempt=1

while attempt<=3:
    print('Attempt',attempt)
    attempt+=1
    
readings=[5,12,3,9]
high=[]
for x in readings:
    if x>=8:
        high.append(x)

print(high)
```

### List comprehensions

```python
# List comprehensions Example
squares = [x**2 for x in range(10)]
print("Squares1:", squares)

squares=[]
for x in range(10):
    squares.append(x**2)
print("Squares2:", squares)
```

```python
readings=[5,12,3,9]
print([x for x in readings if x>=8])
```
