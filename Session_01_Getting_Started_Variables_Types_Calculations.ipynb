## Concept 2 — Variables and fundamental types

A variable is a *name for a value*, such as `site_name` or `reading`. Use snake_case. `type()` tells you the type. Strings require quotes. `None` represents missing or unspecified information; it is not the same as zero.

**Predict → run → change:** Before running the next code cell, explain what it should print. Next, change **one** value and observe the result.

```python
site_name = 'Site A'
reading = 17.5
count = 3
is_active = True
last_check = None
for value in [site_name, reading, count, is_active, last_check]:
    print(value, type(value).__name__)
```

## Concept 3 — Calculations and safe conversion

Operators: `+ - * / // % **`. Parentheses clarify order. User input and CSV fields are often **strings**, even when they look numeric. Convert deliberately with `int()` and `float()`. Division `/` returns a float.

**Predict → run → change:** Before running the next code cell, explain what it should print. Next, change **one** value and observe the result.

```python
hours = 6
rate = 32.5
total = hours * rate
print('Total:', total)
print('Hours from text:', int('6'))
print('Rounded:', round(total / 7, 2))
```

**Stop and think:** Find the difference between `/` and `//` for 13 divided by 4.
