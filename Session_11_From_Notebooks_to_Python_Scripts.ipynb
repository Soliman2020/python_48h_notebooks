## 1. Notebook code and script code

A notebook is useful for exploration: code, output, notes, and charts can appear together. A Python script is a saved `.py` file that can be run repeatedly from a terminal, scheduled, tested, or imported by another file.

The same Python statement can appear in either place:

```python
message = "Hello from Python"
print(message)
```

The difference is usually the workflow around the code: where it is saved, how it is run, and whether other files can reuse it.

````python
import sys
from pathlib import Path

print("Notebook interpreter:", sys.executable)
print("Python version:", sys.version.split()[0])
print("Current folder:", Path.cwd())
````

## 2. Functions and the `main()` pattern

Top-level code runs immediately when a file is imported. Put the program workflow inside `main()` so the file can be imported without starting the program unexpectedly.

```python
def main():
    print("Program started")


if __name__ == "__main__":
    main()
```

When Python executes a file directly, `__name__` is `"__main__"`. When another file imports it, `__name__` is the module name instead.

````python
def calculate_total(price, quantity):
    """Return price multiplied by quantity."""
    return price * quantity


def main():
    total = calculate_total(12.50, 3)
    print(f"Total: {total:.2f}")


if __name__ == "__main__":
    main()
````

## 4. Command-line arguments with `argparse`

Interactive `input()` is useful for learning. Command-line arguments are useful when a script is automated or called by another process. Python's standard library includes `argparse`.

Save this as `greet.py` and run `python greet.py --name Ada`:

```python
import argparse


def main():
    parser = argparse.ArgumentParser()
    parser.add_argument("--name", default="student")
    args = parser.parse_args()
    print(f"Hello, {args.name}!")


if __name__ == "__main__":
    main()
```

## 5. Portable paths with `pathlib`

Avoid building paths by joining strings manually. `pathlib.Path` works across operating systems and makes file operations readable.

```python
from pathlib import Path

data_path = Path("datasets") / "messy_work_items.csv"
print(data_path)
print(data_path.exists())
```

The path is relative to the current working directory. A script should either document the expected working directory or accept paths as command-line arguments.

````python
from pathlib import Path

project_root = Path.cwd()
print("Current folder:", project_root)
print("Python files here:", sorted(path.name for path in project_root.glob("*.py")))
````

## 6. Move a Pandas workflow into modules

A notebook is a good place to discover a cleaning workflow. Once the decisions are understood, reusable code can move into a module.

`work_items.py`:

```python
import pandas as pd


def load_work_items(path):
    return pd.read_csv(path)


def clean_work_items(data):
    result = data.drop_duplicates(subset="case_id").copy()
    result["hours_open"] = pd.to_numeric(
        result["hours_open"],
        errors="coerce",
    )
    return result
```

`main.py`:

```python
from pathlib import Path

from work_items import clean_work_items, load_work_items


def main():
    data_path = Path("datasets") / "messy_work_items.csv"
    raw = load_work_items(data_path)
    clean = clean_work_items(raw)
    print("Rows before cleaning:", len(raw))
    print("Rows after cleaning:", len(clean))


if __name__ == "__main__":
    main()
```

## 7. Virtual environments

A virtual environment gives one project its own Python packages. This reduces conflicts between projects.

Create one from a terminal in your project folder:

```text
python -m venv .venv
```

Windows PowerShell activation:

```text
.venv\Scripts\Activate.ps1
```

Windows Command Prompt activation:

```text
.venv\Scripts\activate.bat
```

macOS or Linux activation:

```text
source .venv/bin/activate
```

Then check the interpreter and install only what the project needs:

```text
python --version
python -m pip --version
python -m pip install pandas
```

Do not commit `.venv` to a repository. Add it to `.gitignore`.

## 8. Small checks and tests

A script is easier to trust when reusable functions have checks. Begin with a plain `assert`:

```python
assert calculate_total(10, 3) == 30
```

A project can later use `pytest` or the standard-library `unittest`. The important habit is to make expected behavior executable.

````python
def clean_name(name):
    return name.strip().title()


# Small self-checks are executable documentation.
assert clean_name("  ada lovelace  " ) == "Ada Lovelace"
assert calculate_total(10, 3) == 30
print("Self-checks passed.")
````

## Common mistakes and next steps

- Save the file before running it.
- Check the terminal's current working directory.
- Use the same interpreter and virtual environment for installation and execution.
- Keep imports at the top of the module.
- Protect executable workflow code with the `__main__` guard.
- Prefer `Path` objects over hand-built platform-specific strings.
- Do not commit `.venv`, passwords, tokens, or private data.
- Read a script before running unfamiliar setup or installation commands.
