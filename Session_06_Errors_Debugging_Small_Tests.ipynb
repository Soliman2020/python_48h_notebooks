## Concept 1 — Common errors and tracebacks

Python reports an error type and where it occurred. `NameError`: unknown variable; `TypeError`: wrong operation for types; `ValueError`: conversion problem; `KeyError`: missing dictionary key. Debug by reproducing a small failing input and inspecting types.

## Concept 2 — Validate inputs and raise useful errors

A function should check assumptions before computing. `raise ValueError(...)` explains invalid input clearly. Do not silence every exception with `except:`. For beginner exercises, first validate non-empty inputs and allowed ranges.

**Predict → run → change:** Before running the next code cell, explain what it should print. Next, change **one** value and observe the result.

```python
def utilization(used,capacity):
    if capacity<=0:
        raise ValueError('capacity must be greater than zero')
    if used<0:
        raise ValueError('used cannot be negative')
    return round(used/capacity*100,1)
print(utilization(30,40))
```

## Concept 3 — Testing and systematic debugging

A test consists of input and expected output. Include normal, boundary and invalid cases. `assert actual == expected` checks the result. For workflows, inspect one intermediate variable rather than printing everything.

**Predict → run → change:** Before running the next code cell, explain what it should print. Next, change **one** value and observe the result.

```python
def categorize(hours):
    return 'late' if hours>24 else 'on_time'
assert categorize(0)=='on_time'
assert categorize(24)=='on_time'
assert categorize(25)=='late'
print('Three test cases passed')
```
