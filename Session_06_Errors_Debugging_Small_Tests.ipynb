## Concept 1 — Types of errors and tracebacks

Python problems usually fall into three groups:

1. **Syntax errors:** Python cannot understand the structure of the code.
2. **Runtime errors (exceptions):** The code is valid, but something goes wrong while it runs.
3. **Logic errors:** The program runs, but the result is not what you intended.

Common exceptions include `NameError`, `TypeError`, `ValueError`, `KeyError`, `IndexError`, and `ZeroDivisionError`. Read the last line of a traceback first; it usually names the error type.

| Exception | When it happens | Example code | Error message |
|---|---|---|---|
| `NameError` | Using a variable or function that was never defined | `print(undefined_variable)` | `name 'undefined_variable' is not defined` |
| `TypeError` | Applying an operation to the wrong data type | `len(5)` | `object of type 'int' has no len()` |
| `ValueError` | Correct type, but invalid value/content | `int('abc')` | `invalid literal for int() with base 10: 'abc'` |
| `KeyError` | Accessing a dictionary key that doesn't exist | `d = {'a': 1}; d['b']` | `'b'` |
| `IndexError` | Accessing a list index beyond its length | `lst = [1, 2, 3]; lst[5]` | `list index out of range` |
| `ZeroDivisionError` | Dividing a number by zero | `10 / 0` | `division by zero` |

The next code cell handles two predictable input problems. Predict the output before running it.

```python
examples = ['25', 'twenty-five', None]

for item in examples:
    try:
        number = int(item)
    except (TypeError, ValueError) as error:
        print(f'Could not convert {item!r}: {type(error).__name__}')
    else:
        print(f'Converted successfully: {number}')
```

A logic error does not stop the program. The next example runs, but one calculation is intentionally wrong.

```python
price = 80
quantity = 3
wrong_total = price + quantity
right_total = price * quantity

print('Wrong total:', wrong_total)
print('Right total:', right_total)
assert right_total == 240
```

### `try`, `except`, `else`, and `finally`

- `try` contains code that may fail.
- `except` responds to a specific error.
- `else` runs only when `try` succeeds.
- `finally` runs whether an error occurs or not.

Both `else` and `finally` are optional. Use `finally` for a completion message or cleanup, not to hide an error.

```python
text = '42'

try:
    number = int(text)
except ValueError:
    print('Please enter a whole number.')
else:
    print('Conversion succeeded:', number)
finally:
    print('Conversion attempt finished.')
```

## Concept 2 — Validate inputs and raise useful errors

A function should check its assumptions before computing. `raise ValueError(...)` gives the caller a clear explanation instead of returning a misleading result.

```python
def percentage(part, total):
    if total <= 0:
        raise ValueError('total must be greater than zero')
    if part < 0 or part > total:
        raise ValueError('part must be between 0 and total')
    return round(part / total * 100, 1)

print(percentage(3, 4))
```

The next example validates text before using it. This pattern is useful for names, labels, categories, and form fields.

```python
def clean_label(label):
    """
    Cleans a label by stripping whitespace and converting it to title case.

    Filter:
    - not isinstance(label, str) — reject anything that isn't a string (an int, None, a list...). 
    - not label.strip() — only runs when label is definitely a str. strip() removes all leading/trailing whitespace; 
      if nothing remains, the string was empty or whitespace-only, and not "" is True.
    """
    if not isinstance(label, str) or not label.strip():
        raise ValueError('label must be a non-empty text value')
    return label.strip().title()

print(clean_label('  morning shift  '))
```

The caller can catch a validation error and choose what to display.

```python
try:
    result = percentage(8, 4)
except ValueError as error:
    print('Could not calculate percentage:', error)
else:
    print('Percentage:', result)
```

## Concept 3 — Testing and systematic debugging

A useful test includes an input and an expected result. Test a normal case, a boundary case, and an invalid case. When a result is surprising, inspect one small intermediate value instead of printing everything.

```python
def delivery_label(days):
    if days < 0:
        raise ValueError('days cannot be negative')
    if days <= 3:
        return 'standard'
    else:
        return 'late'

assert delivery_label(0) == 'standard'
assert delivery_label(3) == 'standard'
assert delivery_label(4) == 'late'
print('Delivery tests passed')
```

```python
# Another way to write the delivery_label function using a conditional expression:

def delivery_label(days):
    if days < 0:
        raise ValueError('days cannot be negative')
    return 'standard' if days <= 3 else 'late'

assert delivery_label(0) == 'standard'
assert delivery_label(3) == 'standard'
assert delivery_label(4) == 'late'
print('Delivery tests passed')
```

```python
try:
    delivery_label(-1)
except ValueError:
    print('Invalid input test passed')
else:
    raise AssertionError('Negative days should raise ValueError')
```
