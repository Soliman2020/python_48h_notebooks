## Concept 1 — Lists and tuples

A list is ordered and mutable; a tuple is ordered and immutable. Use lists for changing batches of measurements. Use tuples for fixed coordinates or constant paired values. Use `append()`, `len()`, indexing, and slicing.

**Predict → run → change:** Before running the next code cell, explain what it should print. Next, change **one** value and observe the result.

```python
readings=[23.0,24.5,22.5]
readings.append(25.0)
print(readings[0],readings[-1],readings[:2])
location=(29.97,31.14)
print(location)
```

**Stop and think:** Why can we append to a list but not a tuple?

## Concept 2 — Sets and duplicates

Sets keep unique hashable values and do not promise an indexing order. Convert a list to a set to check unique items or use set operations to compare ID lists. Do not rely on a set retaining its display order.

**Predict → run → change:** Before running the next code cell, explain what it should print. Next, change **one** value and observe the result.

```python
ids=['A1','A2','A1','A3','A2']
unique=set(ids)
print('records:',len(ids),'unique:',len(unique))
team_a={'A1','A3'}
team_b={'A2','A3'}
print('shared:',team_a&team_b)
```

**Stop and think:** How many duplicate occurrences are there?

**Industry transfer:** The logic above can be applied to auditor case registry; telecom identifiers; lab sample metadata; employee records. What is the equivalent of our variables or records in your work?

## Concept 3 — Dictionaries and nested records

A dictionary maps keys to values. Use `record.get("key", default)` when fields may be missing. Real CSV or JSON records often resemble a list of dictionaries, which later becomes a DataFrame.

**Predict → run → change:** Before running the next code cell, explain what it should print. Next, change **one** value and observe the result.

```python
ticket={'id':'T-01','status':'open','priority':'high'}
ticket['owner']='Mona'
print(ticket.get('status'))
print(ticket.get('duration',0))
records=[ticket, {'id':'T-02','status':'closed','priority':'low'}]
print(records[1]['id'])
```
