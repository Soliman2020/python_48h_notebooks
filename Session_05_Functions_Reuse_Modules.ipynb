### Lambda functions

````python
# A regular function to add two numbers
def add(x, y):
    return x + y

# Equivalent lambda function
add_lambda = lambda x, y: x + y

# Using the regular function
print(add(3, 5))  # Output: 8

# Using the lambda function
print(add_lambda(3, 5))  # Output: 8
````

def LongestWord(sentence):

    # Split the sentence into words

    # Initialize variables to keep track of the longest word
    longest_word = ""
    
    # Iterate through each word in the list

        if len(word) > len(longest_word):

## Concept 2 — Return, defaults and scope

`return` passes a value to the caller; `print` only displays it. A default parameter provides a sensible fallback. Variables created inside functions usually do not exist outside them. Write a docstring to explain what the function does.

**Predict → run → change:** Before running the next code cell, explain what it should print. Next, change **one** value and observe the result.

````python
def is_overdue(hours,limit=24):
    """Return True when hours exceeds the allowed limit."""
    return hours>limit
flag=is_overdue(30)
print(flag, is_overdue(30,48))
````

````python
def add_and_return(a:int, b:int) -> int:
    result = a + b
    return result  # returns the result to the caller

def add_and_print(a, b):
    result = a + b
    print(result)  # prints the result to the console

x = add_and_return(a=4, b=5)
x
````

````python
# Using the functions
returned_value = add_and_return(3, 5)
print("Returned value:", returned_value+5)  # Output: Returned value: 13

printed_value = add_and_print(3, 5)  # Output: 8
print("Printed value:", printed_value+5)  # Output: Printed value: ?????

add_and_return(3, 5) # returns the sum (8), which can be assigned to a variable and used later.
add_and_print(3, 5) # prints the sum (8) to the console, but it does not return anything. Therefore, printed_value is None.
````

## Concept 3 — Libraries and small test cases

`import` lets you reuse existing Python code. The standard library includes `statistics`, `math`, `datetime`, `json` and `pathlib`. `assert` is a compact check for assumptions while learning; it does not replace formal production testing.

**Predict → run → change:** Before running the next code cell, explain what it should print. Next, change **one** value and observe the result.

````python
import statistics
readings=[20,21,22,30]
print('Mean:',statistics.mean(readings))
print('Median:',statistics.median(readings))

def format_case(case_id,status):
    return f'{case_id}: {status.upper()}'

assert format_case('C-01','open')=='C-01: OPEN'
print('Self-check passed')
````
